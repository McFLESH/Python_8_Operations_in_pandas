```python
# Измените таблицу, полученную из входного CSV-файла, так, чтобы в ней остались лишь те строки,
# которые имеют указанный в качестве второго параметра функции континент (фильтрация по столбцу "continent").
# Заполните строки, имеющие пропущенное значение, минимальным значением по этому столбцу, используя функции fillna() и min().
# Сохраните итоговую таблицу в файл формата CSV с названием "CountryOutput.csv". В результате работы функция должна вернуть получившийся датафрейм.
import pandas as pd

def countries_in_continent(fael, continent):
  data = pd.read_csv(fael)
  data = data[data['continent'] == continent]
  data = data.fillna(data.min())
  data.to_csv('CountryOutput.csv')
  print(data.isnull().sum())
  return data
print(countries_in_continent('CountryTable.csv', 'Europe'))
```
