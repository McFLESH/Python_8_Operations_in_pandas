```python
# Объявите функцию make_dataframe, которая принимает на вход две переменные - словарь (или список словарей),
# а также список индексов. В итоге возвращается датафрейм из полученных данных.
import pandas as pd
def make_dataframe(data, indexes):
    F = pd.DataFrame(data = data, index = indexes)
    return F
```
