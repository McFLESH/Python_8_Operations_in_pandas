```python
# Объявите функцию val_counts, которая принимает на вход две переменные - датафрейм и строку (название столбца).
# В итоге возвращается количество уникальных совпадений в соответствующей колонке.
import pandas as pd
def val_counts(data, clmn):
    k = data[clmn].value_counts(normalize=True)
    return k
```
