```python
# Объявите функцию max_min_column, которая принимает на вход две переменные - датафрейм и строку (название столбца).
# В итоге возвращается минимальное и максимальное значение в заданном столбце
import pandas as pd
def max_min_column(df, clmn):
    col_max = df[clmn].max()
    col_min = df[clmn].min()
    return f'Max = {col_max}, min = {col_min}'
```
